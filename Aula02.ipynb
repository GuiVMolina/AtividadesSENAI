```python
numero = int(input("Digite um número inteiro: "))

if numero >= 0:
  print(f"O número {numero} é maior ou igual a zero (positivo)")
else:
  print(f"O número {numero} é menor que zero (negativo)")
```

```python
numero = int(input("Digite um número inteiro: "))

if numero % 2 == 0:
  print(f"O número {numero} é par")
else:
  print(f"O número {numero} é ímpar")
```

```python
idade = int(input("Qual a sua idade? "))

if idade >= 16:
  print(f"Você já pode votar!")
elif 0 < idade < 16:
  print(f"Você precisa esperar mais {16 - idade} ano(s) para poder votar")
else:
  print(f"Opção inválida")
```

```python
temperatura = float(input("Qual a temperatura atual (em Celsius)? "))

if temperatura >= 30:
  print(f"Está muito quente, beba água ({temperatura:.1f}°C)")
elif 18 < temperatura < 30:
  print(f"A temperatura está agradável ({temperatura:.1f}°C)")
else:
  print(f"Está muito frio, vista-se bem ({temperatura:.1f}°C)")
```

```python
valor = float(input("Valor da compra: R$"))

if valor > 150:
  print(f"O cliente recebe frete grátis")
elif 0 < valor <= 150:
  print(f"O cliente vai precisar pagar o frete")
else:
  print(f"Opção inválida")
```

```python
sinal = input("O farol está no sinal... ").lower

if sinal == "vermelho":
  print(f"Pare!")
elif sinal == "verde":
  print(f"Avance")
elif sinal == "amarelo":
  print(f"Tome cuidado, desacelere")
else:
  print(f"Opção inválida")
```
